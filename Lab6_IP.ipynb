# Lab 6 – Image Smoothing

Smooth the grayscale input image using an unweighted averaging filter, the specified weighted filter, and a Gaussian filter.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = '/content/pexels-chevanon-1108099.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError('Image not found. Check the image path.')

plt.imshow(image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')
plt.show()
```

## Enter Filter Parameters

Enter an odd filter size, the Gaussian sigma, and the padding method.

```python
k = int(input('Enter filter size (odd number such as 3 or 5): '))
sigma = float(input('Enter sigma for Gaussian filter: '))
padding_method = input('Enter padding method (zero / replicate): ').lower()

if k < 3 or k % 2 == 0:
    raise ValueError('Filter size must be an odd number greater than or equal to 3.')
if sigma <= 0:
    raise ValueError('Sigma must be greater than zero.')
if padding_method not in ['zero', 'replicate']:
    raise ValueError('Choose zero or replicate padding.')

if padding_method == 'zero':
    border_type = cv2.BORDER_CONSTANT
else:
    border_type = cv2.BORDER_REPLICATE
```

## Create the Three Filters

The weighted filter follows the given expression: `h(x,y) = max(|x|, |y|)`. Each filter is normalized so that its weights add up to 1.

```python
# Averaging filter
averaging_filter = np.ones((k, k), dtype=np.float32) / (k * k)

# Weighted filter: h(x,y) = max(|x|, |y|)
weighted_filter = np.zeros((k, k), dtype=np.float32)
center = k // 2
for i in range(k):
    for j in range(k):
        x = i - center
        y = j - center
        weighted_filter[i, j] = max(abs(x), abs(y))
weighted_filter = weighted_filter / np.sum(weighted_filter)

# Gaussian filter
gaussian_filter = np.zeros((k, k), dtype=np.float32)
for i in range(k):
    for j in range(k):
        x = i - center
        y = j - center
        gaussian_filter[i, j] = np.exp(-(x*x + y*y) / (2 * sigma*sigma))
gaussian_filter = gaussian_filter / np.sum(gaussian_filter)

print('Averaging Filter:\n', averaging_filter)
print('Weighted Filter:\n', weighted_filter)
print('Gaussian Filter:\n', gaussian_filter)
```

## Apply the Filters and Display the Results

```python
average_image = cv2.filter2D(image, -1, averaging_filter, borderType=border_type)
weighted_image = cv2.filter2D(image, -1, weighted_filter, borderType=border_type)
gaussian_image = cv2.filter2D(image, -1, gaussian_filter, borderType=border_type)

plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(average_image, cmap='gray')
plt.title('Averaging Filter')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(weighted_image, cmap='gray')
plt.title('Weighted Filter')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(gaussian_image, cmap='gray')
plt.title('Gaussian Filter')
plt.axis('off')
plt.show()
```
