# Lab 5 – Histogram Equalization and Histogram Matching

Use the grayscale dog image to perform histogram equalization and histogram matching with a reference image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = '/content/pexels-chevanon-1108099.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError('Image not found. Check the image path.')

plt.imshow(image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')
plt.show()
```

## 1. Histogram Equalization

Calculate the histogram and cumulative distribution function (CDF), then use the CDF to enhance the contrast of the image.

```python
histogram, bins = np.histogram(image.flatten(), 256, [0, 256])
cdf = histogram.cumsum()
cdf_normalized = (cdf - cdf.min()) * 255 / (cdf.max() - cdf.min())

equalized_image = np.interp(image.flatten(), range(256), cdf_normalized)
equalized_image = equalized_image.reshape(image.shape).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis('off')
plt.show()
```

## 2. Histogram Matching

Match the histogram of the input image with the histogram of a reference image. The bundled `reference_gray.jpg` is a brighter version of the dog image so the notebook can run directly. Replace it with a reference image supplied by your instructor, if one is specified.

```python
reference_path = '/content/reference_gray.jpg'
reference_image = cv2.imread(reference_path, cv2.IMREAD_GRAYSCALE)

if reference_image is None:
    raise FileNotFoundError('Reference image not found. Upload reference_gray.jpg.')

hist_input, bins = np.histogram(image.flatten(), 256, [0, 256])
hist_reference, bins = np.histogram(reference_image.flatten(), 256, [0, 256])

cdf_input = hist_input.cumsum() / hist_input.sum()
cdf_reference = hist_reference.cumsum() / hist_reference.sum()

lookup_table = np.zeros(256, dtype=np.uint8)
for i in range(256):
    lookup_table[i] = np.argmin(np.abs(cdf_reference - cdf_input[i]))

matched_image = lookup_table[image]

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Input Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(reference_image, cmap='gray')
plt.title('Reference Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(matched_image, cmap='gray')
plt.title('Histogram Matched Image')
plt.axis('off')
plt.show()
```
