# Lab 7 – Image Sharpening

Sharpen the input image using the Laplacian filter, Sobel and Canny edge operators, and unsharp masking.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = '/content/pexels-chevanon-1108099.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError('Image not found. Check the image path.')

plt.imshow(image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')
plt.show()
```

## 1. Sharpening using Laplacian Filter

```python
laplacian = cv2.Laplacian(image, cv2.CV_64F)
sharpened_laplacian = np.clip(image.astype(np.float64) - laplacian, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sharpened_laplacian, cmap='gray')
plt.title('Laplacian Sharpened Image')
plt.axis('off')
plt.show()
```

## 2. Sharpening using Sobel and Canny Edge Detector

```python
sobel_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)
sobel_magnitude = np.sqrt(sobel_x**2 + sobel_y**2)
sobel_sharpened = np.clip(image.astype(np.float64) + 0.5 * sobel_magnitude, 0, 255).astype(np.uint8)

canny_edges = cv2.Canny(image, 100, 200)
canny_sharpened = np.clip(image.astype(np.float64) + 0.5 * canny_edges, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.imshow(sobel_magnitude, cmap='gray')
plt.title('Sobel Gradient')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(sobel_sharpened, cmap='gray')
plt.title('Sobel Sharpened Image')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(canny_edges, cmap='gray')
plt.title('Canny Edges')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(canny_sharpened, cmap='gray')
plt.title('Canny Sharpened Image')
plt.axis('off')
plt.show()
```

## 3. Unsharp Masking

The Gaussian blurred image must use a 5×5 filter and sigma = 3.

```python
blurred_image = cv2.GaussianBlur(image, (5, 5), 3)
mask = image.astype(np.float64) - blurred_image.astype(np.float64)
unsharp_image = np.clip(image.astype(np.float64) + mask, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(blurred_image, cmap='gray')
plt.title('Gaussian Blur (5x5, sigma=3)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(unsharp_image, cmap='gray')
plt.title('Unsharp Masking')
plt.axis('off')
plt.show()
```
