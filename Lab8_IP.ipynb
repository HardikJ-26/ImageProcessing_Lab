# Lab 8 – Noise Distributions

Generate Uniform, Gaussian, Erlang, Exponential, and Rayleigh noise using their equations. Compare each with the corresponding NumPy random function and plot the noise histograms.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = '/content/pexels-chevanon-1108099.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError('Image not found. Check the image path.')

rows, cols = image.shape
```

## Generate the Noise using Equations and Inbuilt Functions

Python does not have MATLAB's `imnoise` function. The equivalent built-in random distribution functions from NumPy are used here for comparison.

```python
# 1. Uniform Noise
a = -30
b = 30
uniform_noise = a + (b - a) * np.random.rand(rows, cols)
uniform_builtin = np.random.uniform(a, b, (rows, cols))

# 2. Gaussian Noise using Box-Muller equation
u1 = np.random.rand(rows, cols) + 1e-7
u2 = np.random.rand(rows, cols)
gaussian_noise = 25 * np.sqrt(-2 * np.log(u1)) * np.cos(2 * np.pi * u2)
gaussian_builtin = np.random.normal(0, 25, (rows, cols))

# 3. Erlang Noise
k = 3
alpha = 0.2
product = np.ones((rows, cols))
for i in range(k):
    product = product * (np.random.rand(rows, cols) + 1e-7)
erlang_noise = -(1 / alpha) * np.log(product) - (k / alpha)
erlang_builtin = np.random.gamma(k, 1 / alpha, (rows, cols)) - (k / alpha)

# 4. Exponential Noise
alpha_exp = 0.1
exponential_noise = -(1 / alpha_exp) * np.log(1 - np.random.rand(rows, cols) + 1e-7) - (1 / alpha_exp)
exponential_builtin = np.random.exponential(1 / alpha_exp, (rows, cols)) - (1 / alpha_exp)

# 5. Rayleigh Noise
beta = 500
rayleigh_noise = np.sqrt(-beta * np.log(1 - np.random.rand(rows, cols) + 1e-7))
rayleigh_noise = rayleigh_noise - np.mean(rayleigh_noise)
rayleigh_builtin = np.random.rayleigh(np.sqrt(beta / 2), (rows, cols))
rayleigh_builtin = rayleigh_builtin - np.mean(rayleigh_builtin)
```

## Plot Histograms and Noisy Images

```python
noise_names = ['Uniform', 'Gaussian', 'Erlang', 'Exponential', 'Rayleigh']
formula_noises = [uniform_noise, gaussian_noise, erlang_noise, exponential_noise, rayleigh_noise]
builtin_noises = [uniform_builtin, gaussian_builtin, erlang_builtin, exponential_builtin, rayleigh_builtin]

plt.figure(figsize=(15, 18))
for i in range(5):
    noisy_image = np.clip(image.astype(np.float32) + formula_noises[i], 0, 255).astype(np.uint8)

    plt.subplot(5, 3, 3*i + 1)
    plt.hist(formula_noises[i].ravel(), bins=50)
    plt.title(noise_names[i] + ' Noise (Equation)')

    plt.subplot(5, 3, 3*i + 2)
    plt.hist(builtin_noises[i].ravel(), bins=50)
    plt.title(noise_names[i] + ' Noise (NumPy)')

    plt.subplot(5, 3, 3*i + 3)
    plt.imshow(noisy_image, cmap='gray')
    plt.title(noise_names[i] + ' Noisy Image')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
